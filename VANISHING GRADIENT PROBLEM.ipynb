```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import tensorflow as tf
import keras
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from keras.models import Sequential
from keras.layers import Dense
```

```python
X,Y =make_moons(n_samples=1000,noise=0.1,random_state=42)
```

```python
plt.scatter(X[:,0],X[:,1],c=Y,cmap='coolwarm',edgecolors='k')
```

```python
model = Sequential()

model.add(Dense(10,activation='sigmoid',input_dim=2))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(10,activation='sigmoid'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
old_weight = model.get_weights()[0]
print(old_weight)
```

```python
model.fit(X,Y,epochs=10)
```

```python
new_weights = model.get_weights()[0]
print(new_weights)
```

```python
percent = abs((new_weights-old_weight)/old_weight)*100
print(percent)
```

To Solve The Vanishing Problem

1) Reduce The Depth Neural Network
2) Use "Relu" in Hidden layer insted of "Sigmoid"

```python
# Reducing The Depth Of the Model

model1 = Sequential()

model1.add(Dense(10,activation='sigmoid',input_dim=2))
model1.add(Dense(10,activation='sigmoid'))
model1.add(Dense(10,activation='sigmoid'))
model1.add(Dense(1,activation='sigmoid'))
```

```python
model1.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model1.fit(X,Y,epochs=10)
```

```python
new_weight1 = model1.get_weights()[0]
print(new_weight1)
```

```python
percent1 = abs((new_weight1-old_weight)/old_weight)*100
print(percent1)
```

```python
# Use Relu in hidden layer insted of sigmoid

model2 = Sequential()

model2.add(Dense(10,activation='relu',input_dim=2))
model2.add(Dense(10,activation='relu'))
model2.add(Dense(10,activation='relu'))
model2.add(Dense(1,activation='sigmoid'))
```

```python
model2.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
old_weight2 = model2.get_weights()[0]
print(old_weight2)
```

```python
model2.fit(X,Y,epochs=50)
```

```python
percnt2 = abs((model2.get_weights()[0]-old_weight2)/old_weight2)*100
print(percnt2)
```

```python
y_pred = model2.predict(X)
print(y_pred)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
print(confusion_matrix(Y,y_pred))
```

```python
print(classification_report(Y,y_pred))
```

```python
print(accuracy_score(Y,y_pred))
```
